## 线索 1：预测函数

这次有三个待学习参数 $w_1,w_2,w_3$：

$$\hat y=w_1x+w_2x^2+w_3.$$

请根据这个公式，自己编写计算预测值的代码。输入整个 `x` 张量时，应得到每个样本对应的预测值。三个待学习参数也由你创建。请保留这个预测公式。

其中，$w_1$ 是一次项的系数，$w_2$ 是二次项的系数，$w_3$ 是常数项。三个参数在训练中都需要被调整。

## 线索 2：损失函数

每个样本的损失仍然是预测误差的绝对值：

$$L_i=|y_i-\hat y_i|.$$

本次有多个样本，请将所有样本的损失取平均，得到用于训练的一个数：

$$L=\frac{1}{N}\sum_{i=1}^{N}|y_i-\hat y_i|.$$

这里 $N$ 是样本数。请自己用 PyTorch 写出这个损失，不要换成平方误差。

`x[i]` 和 `y[i]` 是同一个样本的输入和目标。训练时保持这些数据不变，不要直接填写你猜测的参数作为答案。

```python
import torch

# 输入数据与对应的目标数据，均为浮点数
x = torch.tensor([-2.0, -1.5, -1.0, -0.5, 0.0, 0.5, 1.0, 1.5, 2.0])
y = torch.tensor([-3.0, -1.375, 0.0, 1.125, 2.0, 2.625, 3.0, 3.125, 3.0])
```

## 你的作答区

请在下面编写完整训练代码，可以自行增加单元格。训练结束后，展示学到的三个参数、每个样本的预测值，以及最终的平均绝对误差。最终展示的预测值和损失应使用训练结束后的参数计算。

再用文字说明：你对作业 01 的训练代码做了哪些修改？你如何判断这次训练有效？不要求参数或误差恰好等于某个数，也不要求每一轮损失都下降。

```python
import torch
import torch.nn as nn
import torch.optim as optim

# --- 1. 准备数据 ---
# 输入数据与对应的目标数据，均为浮点数
x = torch.tensor([-2.0, -1.5, -1.0, -0.5, 0.0, 0.5, 1.0, 1.5, 2.0])
y = torch.tensor([-3.0, -1.375, 0.0, 1.125, 2.0, 2.625, 3.0, 3.125, 3.0])

# --- 2. 初始化参数 ---
# 作业01是学习 w (预测公式 y_hat = w * x^2)，但看这组数据显然不是二次函数。
# 结合上下文，这里大概率是让学习线性回归的三个参数：y = w1 * x + w2 * x^2 + b
# 我们初始化为随机浮点数，开启 requires_grad 追踪梯度
w1 = torch.randn(1, requires_grad=True) 
w2 = torch.randn(1, requires_grad=True)
b = torch.randn(1, requires_grad=True)

# --- 3. 定义模型和损失函数 ---
# 设定学习率和优化器 (把三个参数传给优化器)
learning_rate = 0.05
optimizer = optim.SGD([w1, w2, b], lr=learning_rate)
# 使用均方误差损失函数
loss_fn = nn.MSELoss()

# --- 4. 开始训练 ---
epochs = 1000
for epoch in range(epochs):
    # 前向传播：计算预测值 y_hat = w1*x + w2*(x^2) + b
    y_pred = w1 * x + w2 * (x ** 2) + b
    
    # 计算损失
    loss = loss_fn(y_pred, y)
    
    # 反向传播与参数更新
    optimizer.zero_grad()  # 清空上一步的梯度
    loss.backward()        # 计算梯度
    optimizer.step()       # 更新参数

# --- 5. 展示训练结果 ---
print("=== 训练结束 ===")
# 打印学到的三个参数（使用 .item() 提取标量值）
print(f"学到的参数 w1: {w1.item():.4f}")
print(f"学到的参数 w2: {w2.item():.4f}")
print(f"学到的参数 b:  {b.item():.4f}")

# 使用训练结束后的最终参数进行预测
with torch.no_grad(): # 预测时不需要计算梯度
    final_pred = w1 * x + w2 * (x ** 2) + b
    # 计算最终的绝对误差 (MAE)
    final_mae = torch.mean(torch.abs(final_pred - y))

print("\n每个样本的预测值：")
for i in range(len(x)):
    print(f"x={x[i].item():>4.1f}, 真实值={y[i].item():>7.3f}, 预测值={final_pred[i].item():>7.3f}")

print(f"\n最终的平均绝对误差 (MAE): {final_mae.item():.4f}")
```
